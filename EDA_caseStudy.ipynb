```python
#data visualization -->matplotlib,seaborn,plotly
import matplotlib.pyplot as plt 
```

```python
#distribution  of age
plt.hist(data['Age'],bins=15,color='pink')
plt.xlabel("age")
plt.ylabel("count")
plt.title("Age Distribution")
plt.show()
```

```python
#display the count on bars in histogram
#distribution of age
counts,edges,bars=plt.hist(data['Age'],bins=15,color='coral')
plt.xlabel("age")
plt.ylabel("count")
plt.bar_label(bars)
plt.title("Age Distribution")
plt.show()
```

```python
#bar plot representation
data['Sex'].value_counts().plot(kind='bar')
```

```python
data['Pclass'].value_counts().plot(kind='barh') # barh -h means horizantal
```

```python
plt.bar(data['Survived'],data['Sex'],color=['aqua'])
plt.xlabel("Survived Rate")
plt.ylabel("Gender")
plt.title("Gender Survived distribution")
plt.show()
```

```python
data['Survived'].value_counts()
```

```python
#549 /(549+342)
index=data['Survived'].value_counts().keys().tolist()
print(index)
labels =data['Survived'].value_counts().values.tolist()
print(labels)
```

```python
#pie chart
plt.pie(x=labels,
        labels=['Dead','Alive'],
        colors=['coral','aqua'],
        autopct="%.1f")
plt.show()
```

```python
#pie chart
plt.pie(x=labels,
        labels=['Dead','Alive'],
        colors=['coral','lightgreen'],
        autopct="%.1f%%",
        shadow=True)
plt.title("Survived distribution")
plt.legend()
plt.show()
```

```python
#pie Chart
plt.pie(x= [577,314],
        labels=['M','F'],
        colors=['orange','lightgreen'],
        autopct="%.1f%%",
       shadow=True)
plt.title("Gender Distribution")
plt.legend()
plt.show()
```

```python
#pairwise relationship -->scatterplot()
plt.scatter(x=data['Age'],y=data['Fare'])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age VS Fare Distribution")
plt.show()
```

```python
#box plot and scatter plot
trace=go.Box(x=data['Pclass'],
            y=data['Age'],
            marker={'color':'purple'})
layout=go.Layout(title="Age vs Pclass Distribution")
fig=go.Figure(data=trace,layout=layout)
fig.show()
```

```python
#scatter plot representation --> Age vs Fare
trace=go.Scatter(x=data['Age'],y=data['Fare'],mode='markers',marker={'color':'black'})
layout=go.Layout(title="Age vs Fare Distribution")
fig=go.Figure(data=trace,layout=layout)
fig.show()
```

```python
#scatter plot representation --> Age vs Fare
trace=go.Scatter(x=data['Age'],y=data['Fare'],marker={'color':'purple'})
layout=go.Layout(title="Age vs Fare Distribution")
fig=go.Figure(data=trace,layout=layout)
fig.show()
```
